```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import tidy3d.web as web

# Load the data through task-id
sim_data = web.load('')

# Select the data you want to plot
plot_data = sim_data["fieldtimemonitor_0"].Ex.sel(x=0).sel(t=1.0323811520601537e-14)

# Plot the data
if plot_data.size == 1:
  print(plot_data)
else:
  fig, ax = plt.subplots()
  plot_data.real.plot(x="y", y="z", ax=ax)
```
